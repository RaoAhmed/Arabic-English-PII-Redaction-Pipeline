```python
from presidio_analyzer import Pattern, PatternRecognizer

# 1. KSA National ID / Iqama (10 digits starting with 1 or 2 in Western 0-9 or Eastern ٠-٩ numerals)

national_id_pattern = Pattern(

    name="ksa_national_id_pattern",

    regex=r"\b[12١٢][0-9٠-٩]{9}\b",

    score=0.85

)



nat_id_en = PatternRecognizer(

    supported_entity="KSA_NATIONAL_ID",

    patterns=[national_id_pattern],

    supported_language="en",

)



nat_id_ar = PatternRecognizer(

    supported_entity="KSA_NATIONAL_ID",

    patterns=[national_id_pattern],

    supported_language="ar",

    context=["هوية", "إقامة", "رقم الهوية", "السجل المدني"]

)



# 2. Saudi Phone Number (+966 or 05 / ٠٥ followed by 8 digits)

phone_pattern = Pattern(

    name="ksa_phone_pattern",

    regex=r"(\+?(966|٩٦٦)|00966|٠٠٩٦٦|0|٠)?[5٥][0-9٠-٩]{8}\b",

    score=0.85

)



phone_en = PatternRecognizer(

    supported_entity="PHONE_NUMBER",

    patterns=[phone_pattern],

    supported_language="en"

)



phone_ar = PatternRecognizer(

    supported_entity="PHONE_NUMBER",

    patterns=[phone_pattern],

    supported_language="ar",

    context=["جوال", "هاتف", "اتصل"]

)



# 3. Saudi IBAN (SA followed by 22 digits)

iban_pattern = Pattern(

    name="ksa_iban_pattern",

    regex=r"\bSA\d{2}[A-Z0-9]{18}\b",

    score=0.95

)



iban_en = PatternRecognizer(

    supported_entity="IBAN_CODE",

    patterns=[iban_pattern],

    supported_language="en"

)



iban_ar = PatternRecognizer(

    supported_entity="IBAN_CODE",

    patterns=[iban_pattern],

    supported_language="ar",

    context=["آيبان", "حساب بكي", "رقم الحساب"]



)



# Arabic Credit Card Recognizer

# Matches 12 to 19 digits (Eastern or Western numerals) grouped by spaces or hyphens.

credit_card_pattern = Pattern(

    name="ar_credit_card_pattern",

    regex=r"(?:\b|\s)[0-9٠-٩]{4,5}[-\s][0-9٠-٩]{4,5}[-\s][0-9٠-٩]{4,5}(?:[-\s][0-9٠-٩]{4,5})?(?:\b|\s)",

    score=0.85

)



cc_ar = PatternRecognizer(

    supported_entity="CREDIT_CARD",

    patterns=[credit_card_pattern],

    supported_language="ar",

    context=["ائتمان", "بطاقة", "فيزا", "ماستركارد", "مدى"]

)



# Driver's License Recognizer

# Matches 9 digits (Eastern or Western numerals) accompanied by context keywords like "رخصة" (license) or "قيادة" (driving).

license_pattern = Pattern(

    name="ar_license_pattern",

    regex=r"(?:\b|\s)[0-9٠-٩]{9}(?:\b|\s)",

    score=0.75

)



license_ar = PatternRecognizer(

    supported_entity="DRIVER_LICENSE",

    patterns=[license_pattern],

    supported_language="ar",

    context=["رخصة", "قيادة", "رخصة قيادة", "سياقة"]

)



# Social Security Number (SSN) Recognizer

# Matches the XXX-XX-XXXX format in Eastern or Western numerals.

ssn_pattern = Pattern(

    name="ar_ssn_pattern",

    regex=r"(?:\b|\s)[0-9٠-٩]{3}-[0-9٠-٩]{2}-[0-9٠-٩]{4}(?:\b|\s)",

    score=0.85

)



ssn_ar = PatternRecognizer(

    supported_entity="US_SSN",

    patterns=[ssn_pattern],

    supported_language="ar",

    context=["الضمان الاجتماعي", "ضمان", "تأمين"]

)



# Polish National ID Recognizer

# Matches 3 uppercase Latin letters followed by 6 digits (ABC123456).

polish_id_pattern = Pattern(

    name="polish_id_pattern",

    regex=r"\b[A-Za-z]{3}\d{6}\b",

    score=0.85

)

polish_id_en = PatternRecognizer(

    supported_entity="POLISH_ID",

    patterns=[polish_id_pattern],

    supported_language="en"

)

polish_id_ar = PatternRecognizer(

    supported_entity="POLISH_ID",

    patterns=[polish_id_pattern],

    supported_language="ar",

    context=["بولندية", "هوية", "بطاقة"]

)



# Arabic Date Pattern (supports Eastern ٠-٩ and Western 0-9 digits)

arabic_months = (

    "يناير|فبراير|مارس|أبريل|مايو|يونيو|يوليو|أغسطس|سبتمبر|أكتوبر|نوفمبر|ديسمبر|"

    "محرم|صفر|ربيع الأول|ربيع الثاني|جمادى الأولى|جمادى الآخرة|رجب|شعبان|رمضان|شوال|ذو القعدة|ذو الحجة"

)



date_pattern = Pattern(

    name="arabic_date_pattern",

    regex=rf"(?:\b|\s)[0-9٠-٩]{{1,2}}\s+(?:{arabic_months})(?:\s+[0-9٠-٩]{{4}})?(?:\b|\s)",

    score=0.85

)



arabic_date_recognizer = PatternRecognizer(

    supported_entity="DATE_TIME",

    patterns=[date_pattern],

    supported_language="ar",

    context=["تاريخ", "بتاريخ", "يوم", "عام", "سنة"]

)



# Arabic Time Pattern (e.g., 9:30 صباحًا / ٠٩:٣٠ مساءً)

time_pattern = Pattern(

    name="arabic_time_pattern",

    regex=r"(?:\b|\s)[0-9٠-٩]{1,2}:[0-9٠-٩]{2}(?:\s*(?:صباحًا|مساءً|ص|م))?(?:\b|\s)",

    score=0.85

)



arabic_time_recognizer = PatternRecognizer(

    supported_entity="DATE_TIME",

    patterns=[time_pattern],

    supported_language="ar",

    context=["ساعة", "الساعة", "تمام", "وقت"]

)
```

```python
import stanza
from presidio_analyzer import RecognizerResult, EntityRecognizer

class NativeArabicStanzaRecognizer(EntityRecognizer):
    def __init__(self, stanza_pipeline=None):
        super().__init__(
            supported_entities=["PERSON", "LOCATION", "ORGANIZATION"],
            supported_language="ar",
            name="native_arabic_stanza_recognizer"
        )

        # Initialize native Stanza pipeline if not provided
        self.pipeline = stanza_pipeline or stanza.Pipeline(
            lang="ar",
            processors="tokenize,mwt,pos,lemma,ner",
            verbose=False,
            # For a Offline environment, Uncomment the following line
            # download_method=None
        )

        self.label_map = {
            "PER": "PERSON",
            "LOC": "LOCATION",
            "ORG": "ORGANIZATION"
        }

    def load(self):
        pass

    def analyze(self, text, entities, nlp_artifacts=None):
        results = []
        # Run Stanza directly on the raw text
        doc = self.pipeline(text)

        for ent in doc.ents:
            entity_type = self.label_map.get(ent.type)
            if entity_type and (not entities or entity_type in entities):
                results.append(
                    RecognizerResult(
                        entity_type=entity_type,
                        start=ent.start_char,
                        end=ent.end_char,
                        score=0.85
                    )
                )
        return results
```

```python
import spacy
from presidio_analyzer.nlp_engine import SpacyNlpEngine
from presidio_analyzer import AnalyzerEngine
from presidio_anonymizer import AnonymizerEngine
from presidio_anonymizer.entities import OperatorConfig

# 1. Create an NlpEngine and register lightweight tokenizers
nlp_engine = SpacyNlpEngine()
nlp_engine.nlp = {
    "en": spacy.load("en_core_web_sm"),
    "ar": spacy.blank("ar"),  # Satisfies Presidio's language check for "ar"
}

# 2. Pass nlp_engine into AnalyzerEngine
analyzer = AnalyzerEngine(
    nlp_engine=nlp_engine,
    supported_languages=["ar", "en"]
)

# Add the native Stanza NER recognizer for names & locations
analyzer.registry.add_recognizer(NativeArabicStanzaRecognizer())

# Add all recognizers to engine registry
analyzer.registry.add_recognizer(nat_id_en)
analyzer.registry.add_recognizer(nat_id_ar)
analyzer.registry.add_recognizer(phone_en)
analyzer.registry.add_recognizer(phone_ar)
analyzer.registry.add_recognizer(iban_en)
analyzer.registry.add_recognizer(iban_ar)

# Add your numeric recognizers (Credit card, Driver license, SSN, Polish ID)
analyzer.registry.add_recognizer(cc_ar)
analyzer.registry.add_recognizer(license_ar)
analyzer.registry.add_recognizer(ssn_ar)
analyzer.registry.add_recognizer(polish_id_ar)
analyzer.registry.add_recognizer(polish_id_en)

# Add Date and Time recognizers
analyzer.registry.add_recognizer(arabic_date_recognizer)
analyzer.registry.add_recognizer(arabic_time_recognizer)

# Custom Operator mappings
operators = {
            "KSA_NATIONAL_ID": OperatorConfig("replace", {"new_value": "[REDACTED_NATIONAL_ID]"}),
            "PHONE_NUMBER": OperatorConfig("replace", {"new_value": "[REDACTED_PHONE]"}),
            "EMAIL_ADDRESS": OperatorConfig("replace", {"new_value": "[REDACTED_EMAIL]"}),
            "PERSON": OperatorConfig("replace", {"new_value": "[REDACTED_NAME]"}),
            "IBAN_CODE": OperatorConfig("replace", {"new_value": "[REDACTED_IBAN]"}),
            "LOCATION":OperatorConfig("replace", {"new_value": "[REDACTED_LOCATION]"}),
            "CREDIT_CARD": OperatorConfig("replace", {"new_value": "[REDACTED_CREDIT_CARD]"}),
            "DRIVER_LICENSE": OperatorConfig("replace", {"new_value": "[REDACTED_LICENSE]"}),
            "DATE_TIME": OperatorConfig("replace", {"new_value": "[REDACTED_DATE_TIME]"}),
            "US_SSN": OperatorConfig("replace", {"new_value": "[REDACTED_SSN]"}),
            "POLISH_ID": OperatorConfig("replace", {"new_value": "[REDACTED_POLISH_ID]"}),
            "DEFAULT": OperatorConfig("replace", {"new_value": "[REDACTED]"}),
        }

anonymizer = AnonymizerEngine()

# Run analysis and anonymization for Arabic Text
results = analyzer.analyze(text=arabic_text, language="ar")

anonymized_result = anonymizer.anonymize(
    text=arabic_text,
    analyzer_results=results,
    operators=operators
)

print(anonymized_result.text)
```
